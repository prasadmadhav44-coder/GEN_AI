### *Scatter plot*

```python
import numpy as np
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])
plt.scatter(x,y,c=["r","g","b","y"],marker="*",s=100)# arrays should be identical
plt.title("Sample dataset")
plt.xlabel("x")
plt.ylabel("y")
plt.show();
```

### *Line Plot*

```python
import numpy as np
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])
plt.plot(x,y,c="r",marker="*",linestyle="dashed")# arrays should be identical
plt.title("Sample dataset 1")
plt.xlabel("x")
plt.ylabel("y")
plt.show();
```

### *Bar plot*

```python
year=[1999,2000,2001,2002,2003]
lang=["None","c","c++","java","python"]
plt.bar(year,lang,color='r',align='edge',edgecolor='black')
plt.show();
```

```python
#frequency Distribution

np.random.seed(4)
a=np.random.normal(50,1,100)
plt.hist(a,color='r',bins=20,edgecolor='black')
plt.show();
```

### *Pie Chart Plot*

```python
lang=["c","c++","java","python"]
votes=[75,80,95,99]
exp=[0,0,0,0.5]
plt.pie(votes,labels=lang,colors=['olivedrab', 'rosybrown', 'gray', 'saddlebrown'],explode=exp,autopct='%1.1f%%',shadow=True,startangle=90)
plt.show();
```

```python
stk1=[1,2,3,4]
stk2=[5,3,7,9]
stk3=[3,5,7,4]
plt.title("Stock Market")
plt.plot(stk1,label='stock1')
plt.plot(stk2,label='stock2')
plt.plot(stk3,label='stock3')
plt.legend()
plt.show();
```

## Subplots Comparing (age vs chol),(age vs thalach)

```python
fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(15,10))
scatter = ax[0].scatter(df["age"],df["chol"],c=df["target"],cmap="winter")
#customise the plot
ax[0].set(title ='Age vs Cholestrol',xlabel="Age",ylabel="Cholestrol")

ax[0].legend(*scatter. legend_elements(), title="Target")

#add mean line
ax[0].axhline(df["chol"].mean(),color="r",linestyle="dashed")

scatter=ax[1].scatter(df["age"],df["thalach"],c=df["target"],cmap="viridis")
#customise the plot
ax[1]. set(title='Age vs Thalach',xlabel="Age",ylabel="Thalach")
ax[1]. legend(*scatter. legend_elements(), title="Target")
ax[1].axhline(df["thalach"].mean(),color="r",linestyle="dashed")

plt.subplots_adjust(hspace=0.40)

plt.show();
```

```python

fig, ax = plt.subplots(figsize=(10, 5))

scatter = ax.scatter(df['age' ], df['trestbps'], c=df['target' ], cmap="winter")

ax. set(title="Heart Disease and Trestbps Levels", xlabel="Age", ylabel="Trestbps")
ax. legend(*scatter. legend_elements(), title='Target')
ax.axhline(df['trestbps'].mean(), linestyle='dashed');

plt.savefig("Heart Disease and Trestbps Levels.png")
plt.show();
```

```python
data = df[df["target"] == 1]["age"].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(data.index, data.values, color="teal")
ax.set(title="Heart Disease Cases by Age",xlabel="Age",ylabel="Number of Cases")
ax.axhline(data.mean(), color="r", linestyle="dashed")

ax.legend(["Average Cases"])
plt.show();
```
