## *Pandas Series*

```python
price = pd.Series([10, 20, 30, 40])
print(price)
```

```python
print(price.dtype)
print(price.shape)
print(price.size)
```

```python
print(price[0])
```

```python
price = pd.Series([10000000, 20000000, 30000000, 4000000000], index=['a', 'b', 'c', 'd'])
print(price)
```

```python
print(price['a'])
```

```python
carname = pd.Series(['Dodge','Ferrari','Bugatti','BMW'],index = ['a', 'b', 'c', 'd'])
print(carname)
```

```python
data = {
    "Car": carname,
    "Price": price
}

df = pd.DataFrame(data)
print(df)
```

```python
df.info()
```
