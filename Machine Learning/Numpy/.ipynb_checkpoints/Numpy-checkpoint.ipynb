## *NumPy - Numerical Python*

```python
import numpy as np 
```

```python
arr1=np.array([10,20,30,40])
print(arr1)
arr1
print(arr1.shape)
print(arr1.size)
```

```python
arr2=np.array([[10,20,30],[10,20,30]])
print(arr2)
arr2
print(arr2.shape)
print(arr2.size)
```

```python
arr3=np.array([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
print(arr3)
arr3
print(arr3.shape)
print(arr3.size)
```
