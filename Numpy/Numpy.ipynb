## *NumPy - Numerical Python*

```python
import numpy as np 
```

```python
arr1=np.array([10,20,30,40])
print(arr1)
arr1
print(arr1.shape)
print(arr1.size)
print(arr1.ndim)
print(arr1.dtype)
```

```python
arr2=np.array([[10,20,30],[10,20,30]])
print(arr2)
arr2
print(arr2.shape)
print(arr2.size)
print(arr2.ndim)
```

```python
arr3=np.array([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
print(arr3)
arr3
print(arr3.shape)
print(arr3.size)
print(arr3.ndim)
```

## *Type casting*

```python
arr1=np.array([10,20,30,"40"],dtype="float")
print(arr1)
arr1

print(arr1.dtype)
```

## *Array Filling*

```python
a=np.full((2,5,4),21)
print(a)
```

```python
a=np.full((2,5,4),[1,2,3,4])
print(a)
```

```python
b=np.ones((5,5),dtype="int")
print(b)
```

```python
c=np.zeros((5,5))
print(c)
```

### Arange

```python
print(np.arange(-10,11,2))
```

```python
print(np.linspace(0,100,500))## split into equal part from 0-100
```

### inf, nan

```python
print(np.sqrt(21))
print(np.sqrt(-21))
print("Hello")
```

```python
print(np.array([10,20,30,40])/2)
print(np.array([10,20,30,40])/0)
print("Hello")
```

```python
print(np.array([10,20,30,40] + np.array([50,60,70,80])))
print(np.array([10,20,30,40] - np.array([50,60,70,80])))
print(np.array([10,20,30,40] * np.array([50,60,70,80])))
```

```python
print(np.array([10,20,30,40] + np.array([[5],[6]])))
```
