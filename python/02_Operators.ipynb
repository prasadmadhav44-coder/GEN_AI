## *operators*

- Arithmetic operators - +, -, *, / , ** ,//,%

- Assignment operators - =, +=, -= , /=, *= , **= , //=, %=

- Comparison operators - == , != , >,<, <= , >=

- Logical operators - and, or, not

- Identity operators - is, is not

- Membership operators - in, not in


```python
#Arithmetic Operators
a=10
b=5
print("Addition: ", a + b)
print("Subtraction: ", a - b)
print("Multiplication: ", a * b)
print("Division: ", a / b)
print("Floor Division: ", a // b)
print("Modulus: ", a % b)
```

```python
#Assignment Operators
a=10
a+=5 # a = a + 5
print("a =", a)
b=10
b-=5 # b = b - 5
print("b =", b)
c=10
c*=5 # c = c * 5
print("c =", c)
d=10
d/=5 # d = d / 5
print("d =", d)
e=10
e//=5 # e = e // 5
print("e =", e)
f=10
f%=5 # f = f % 5
print("f =", f)
```

```python
#comparison Operators
a=10
b=5
print("a == b: ", a == b)#checking if a is equal to b
print("a != b: ", a != b)#checking if a is not equal to b
print("a > b: ", a > b)#checking if a is greater than b
print("a < b: ", a < b)#checking if a is less than b
print("a >= b: ", a >= b)#checking if a is greater than or equal to b
print("a <= b: ", a <= b)#checking if a is less than or equal to b
```

```python
#logical Operators
a=True
b=False
print("a and a: ", a and a)#checking if both a and b are True
print("a and b: ", a and b)#checking if both a and b are False
print("b and a: ", b and a)#checking if both a and b are False
print("b and b: ", b and b)#checking if both a and b are False
print("a or a: ", a or a)#checking if either a or b is True
print("a or b: ", a or b)#checking if either a or b is True
print("b or a: ", b or a)#checking if either a or b is True
print("b or b: ", b or b)#checking if either a or b is False
print("not a: ", not a)#checking if a is False
print("not b: ", not b)#checking if b is True
```

```python
#identity Operators
a=10
b=10
print("a is b: ", a is b)#checking if a and b are the same
print("a is not b: ", a is not b)#checking if a and b are not the same
```

```python
#membership Operators
a=[1,2,3,4,5]
print("1 in a: ", 1 in a)#checking if 1 is in a
print("6 in a: ", 6 in a)#checking if 6 is in a
print("1 not in a: ", 1 not in a)#checking if 1 is not in a
```
