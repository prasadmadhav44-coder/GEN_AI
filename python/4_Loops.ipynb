## *While Loop*
- With the while loop we can execute a set of statements as long as a condition is true.

**Keywords**
- while
- break – stop the loop
- continue – go back to the top of my loop

**syntax**
- while <"condition">:

```python
a=1
while a<=10:
    print(a,end=' ')
    a+=1#jump or increment
print("\nDone")
```

```python
a=0
while a<=100:
    #reverse loop - prints reverse in  100 -0
    print(100-a,end=' ')
    a+=1
    
```

## *Break* - (Breaking the loop Midway)

```python
#Infinite loop
a=1
while True:
    print(a,end=' ')
    a+=1
```

```python
#Break
a=1
while a>=0:
    if a>20:
        break#stops the loop
    print(a,end=' ')
    a+=1
```

## *Continue* -(go back to the top of the loop)

```python
a=1
while a<=20:
    if a%3==0:
        a+=1
        continue # go back to the top of my code222222222222222222
    print(a,end=' ')
    a+=1
    
```

## *Nested While Loop*

```python
a=1
while a<=5:
    b=1
    while b<=5:
        print(b,end='')
        b+=1
    print()
    a+=1
    
```

```python
#pattern
x=int(input("Enter no.of.rows :"))
y=int(input("Enter no.of.colums :"))
a=1
while a<=x:
    b=1
    while b<=y:
        print(b,end='')
        b+=1
    print()
    a+=1
    
```

```python
# printing Symbol Patterns
row=int(input("Enter no.of.rows :"))
coloum=int(input("Enter no.of.colums :"))
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
a=1
while a<=row:
    b=1
    while b<=coloum:
        print(Symbol,end='')
        b+=1
    print()
    a+=1
```
