## *if elif else* (Decision statement)
**syntax**
- if<<condition>>:
- elif<<condition>>:
- else:

```python
a=int(input("Enter a number: "))
if a>100: #: - used for Identation
    pass#pass statement is used to avoid error when there is no code to execute in the if block
if a>100:
    print("a is greater than 100")
elif a==100:
    print("a is equal to 100")
else:
    print("a is less than 100")
print("This is outside the if else block so it will be executed")
```

### *if,elif,else rules*
- if - only once per code block(mandatory)
- elif - as many time as per code block(optional)
- else - only once per code block(optional)

```python
a=int(input("Enter a number: "))
if a>100:
    print("a is greater than 100")
elif a==100:
    print("a is equal to 100")
elif a>50 and a<100:
    print("a is greater than 50 and less than 100")
else:
    print("a is less than 50")
```

## *Nested if & else*

```python
a=int(input("Enter a number: "))
if a>10:
    print("a is greater than 10")
    if a>20:
        print("a is greater than 20")
    else:
        print("a is less than or equal to 20")
elif a<10:
    print("a is less than 10")
else:
    print("a is equal to 10")
    
```

## *Ternary Operator* 

```python
a=int(input("Enter a number: "))
print("a is greater than 10") if a>10 else print("a is less than or equal to 10")#ternary operator is used to write if else statement in a single line
```
