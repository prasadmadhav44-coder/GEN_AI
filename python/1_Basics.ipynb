## *Data Types*
### Primitive
- Integer
- Float
- Complex
- String
- Boolean
- None
### Non-Primitive
- List 
- Tuple
- Dict
- Set

```python
#Integer
a=100#immutable
b=-100
#Float
c=50.55#immutable
d=-50.55
e=2e5 # 200000.0
#String
f="Hello"#immutable
#Boolean
g=True#immutable
h=False
#None
i=None#immutable
#Complex
j=3+5j#immutable
print(a,type(a))
print(b,type(b))
print(c,type(c))
print(d,type(d))
print(e,type(e))
print(f,type(f))
print(g,type(g))
print(h,type(h))
print(i,type(i))
print(j,type(j))
```

```python
#List is a collection which is ordered and changeable. Allows duplicate members.Denoted by square brackets []
a=[1,2,3,4,5]#mutable
#Tuple is a collection which is ordered and unchangeable. Allows duplicate members.Denoted by parentheses ()
b=(1,2,3,4,5)#immutable
#Set is a collection which is unordered and unindexed. No duplicate members.Denoted by curly brackets {}
c={1,2,3,4,5}#mutable
#Dictionary is a collection which is ordered and changeable. No duplicate members.Denoted by curly brackets {Key: Value} with key:value pairs which is Unique.
d={"name":"John","age":30,"city":"New York"}#mutable
print("--list--")
print(a,type(a))
print(a[0])#accessing first element of list using index
print(a[-1])#accessing last element of list using negative index
a[4]=100#changing value of last element of list
print(a,type(a))
print("--tuple--")
print(b,type(b))
print(b[0])#accessing first element of tuple using index
print(b[-1])#accessing last element of tuple using negative index
print("--set--")
print(c,type(c))
print("--dictionary--")
print(d,type(d))
print(d["name"])#accessing value of key "name" in dictionary
print(d["age"])#accessing value of key "age" in dictionary
print(d["city"])#accessing value of key "city" in dictionary
d["age"]=31#changing value of key "age" in dictionary
d["country"]="USA"#adding new key-value pair in dictionary
print(d,type(d))
```

## *Standard Fuctions*

```python
#print() - print() function is used to display the output on the console. It can take multiple arguments and print them on the same line. It can also take optional parameters like sep, end, file, flush etc.
print("Hello, World!")
print(1000 + 2000)
print("Hello", "World", sep="-", end="!\n")#sep - separator, end - end of the line.
print(1,2,3,4,5,end=" ")#end - end of the line.
print(6,7,8,9,10,end="\n")#end - end of the line.
```

## *Escape Sequences*

```python
print("Hello \\World\\") # \\ - prints a single backslash
print("Hello \'World\'") # \' - prints a single quote
print("Hello \"World\"") # \" - prints a double quote
print("Hello \nWorld") # \n - prints a new line
print("Hello \t World") # \t - prints a tab space
# """ - prints a multi-line string
print("""lorem ipsum dolor sit amet, consectetur adipiscing elit, sed do eiusmod tempor incididunt ut labore et dolore magna aliqua.
Ut enim ad minim veniam, quis nostrud exercitation ullamco laboris nisi ut aliquip ex ea commodo consequat. 
Duis aute irure dolor in reprehenderit in voluptate velit esse cillum dolore eu fugiat nulla pariatur.
Excepteur sint occaecat cupidatat non proident, sunt in culpa qui officia deserunt mollit anim id est laborum.""")
```

## *Type Casting*

```python
a=50
print(float(a),type(float(a)))
print(complex(a),type(complex(a)))
print(str(a),type(str(a)))
print(bool(a),type(bool(a)))
b=50.50
print(int(b),type(int(b)))
print(complex(b),type(complex(b)))
print(str(b),type(str(b)))
print(bool(b),type(bool(b)))
c="Hello"
# print(int(c),type(int(c)))#ValueError: invalid literal for int() with base 10: 'Hello'
# print(float(c),type(float(c)))#ValueError: invalid literal for float() with base 10: 'Hello'
# print(complex(c),type(complex(c)))#ValueError: complex() arg is a malformed string
print(bool(c),type(bool(c)))
d=True
print(int(d),type(int(d)))
print(float(d),type(float(d)))
print(str(d),type(str(d)))
print(complex(d),type(complex(d)))
e="500"
print(int(e),type(int(e)))
print(float(e),type(float(e)))
print(str(e),type(str(e)))
print(bool(e),type(bool(e)))# "" 0r 0 is False, any other value is True
```

## *input() - function*

```python
a=input("Enter a number: ")
print("You entered:", a)
b=float(input("Enter a string: "))
print("You entered:", b)
c=int(input("Enter a float: "))
print("You entered:", c)
d=input("Enter a boolean [0 or 1]: ")
print("You entered:", bool(int(d)))
print("b + c = ", b + c)
#input always returns a string, so we need to convert it to the desired type using int(), float(), bool() etc.
```

## Keyword - del

```python
x=input("Enter a number: ")
y=int(x)
print("the value of x is: ",x,type(x))
del x #deleting variable x to free up memory
print("the value of y is: ",y,type(y))
```
