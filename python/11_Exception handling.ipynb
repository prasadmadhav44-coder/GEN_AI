## *Exception Handling*
Try ... Except ... finally

• The try block lets you test a block of code for errors.

• The except block lets you handle the error.

• The finally block lets you execute code, regardless of the result of the  try 
and except blocks.

• The else block works only if try is correct

## *Handling a Exception*

```python
print("Few lines above")
try:
    num = int(input("Enter a number:"))
    print(f"The user enterd number is = {num}")
except:
    print("Error handling is done here")
print("Few lines below")
```

**Dividing a number with Zero Error Handling.**

```python
print("Few lines above")
try:
    num1 = int(input("Enter a Dividend number :"))
    print(f"The User Entered Dividend Number is = {num1}")
    num2 = int(input("Enter a Divisor number :"))
    print(f"The User Entered Divisor Number is = {num2}")
    quotient = num1/num2
    print(f"The Quotient of {num1}/{num2} = {quotient}")
except ValueError:
    print("You gave wrong input type")
except ZeroDivisionError:
    print("Dont divide by zero")
```

## *Finally*
The finally block lets you execute code, regardless of the result of the try and
except blocks

```python
print("Few lines above")
try:
    num1 = int(input("Enter a number :"))
    print(f"The User Entered Number is = {num1}")
except:
    print("Error handling is done here")
finally:
    print("This line will always work")
```

```python
while True:
    try:
        num = int(input("Enter a number inbetween 10-50:"))
        if num<10:
            raise ValueError
        elif num>50:
            raise ZeroDivisionError
        else:
            print(f"Thank you for entering the = {num}")
            break
    except ValueError:
        print("Please Enter a numbaer greater than or equal to 10")
    except ZeroDivisionError:
        print("Please Enter a number lesser than or equal to 50")
```

```python
try:
    num = int(input("Enter a number inbetween 10-50:"))
    if num<10:
        raise ValueError("Please Enter a number greater than or equal to 10")
    elif num>50:
        raise ZeroDivisionError("Please Enter a number lesser than or equal to 50")
    else:
        print(f"Thank you for entering the {num}")
except ValueError as x:
    print(x)
except ZeroDivisionError as y:
    print(y)
```

## *ASSERT - KEYWORD*
Python assert statement is a debugging aid that tests a condition. If the 
condition is true, it does nothing and your program just continues to execute. 
But if the assert condition evaluates to false, it raises an AssertionError
exception with an optional error message.

```python
y = 3
assert y<5
print("hello")
x = 3
assert x>5
print("hello")
```

```python

x = 3
assert x > 5, "my value should be greater than 5"
```

```python

while True:
    try:
        num = int(input("Enter a number between 10-50: "))
        assert num >= 10 and num <= 50, "Please provide a number inbetween 10-50"
        print(f"Thank you for entering the = {num}")
        break
    except Exception as e:
        print(e)
```

```python
try:
    print("Outer try block")
    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print(result)
    except ValueError:
        print("Inner Error: Invalid input (not a number)")
    except ZeroDivisionError:
        print("Inner Error: Cannot divide by zero")
except Exception as e:
    print("Outer Error:", e)
```

```python
try:
    try:
        a = int(input("Enter a: "))
        b = int(input("Enter b: "))
        print(a / b)
    except ZeroDivisionError:
        print("Cannot divide by zero")
except:
    print("Something went wrong")
```
