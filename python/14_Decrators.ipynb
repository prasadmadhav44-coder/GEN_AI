## *Decorator*
To understand decorator we need to understand four concepts

- nested function

- function as a reference

- function can return another function

- function as a parameter

```python
def outer():
    x=10
    def inner():
        y=90
        z=x+y
        return z
    return inner
print(outer()())
print(outer().__name__)
```

```python
def createAdder(x):
    def add(y):
        return x+y
    return add
print(createAdder(6)(3))
ans = createAdder("hello")
print(ans("world"))
```

```python
def function1():
    print("Function 1")
def function2(x):
    x()
    print("Function 2")
function2(function1)
```

```python
#this is written without decorators
# without distrubing this function we are going to add some functionality to it, to do this im going to add additional function.

def strupper(func):
    def inner(): #to modify existing functions functionality
        str1 = func() #strname() is called and str="hi welcome "
        return str1. upper()
    return inner

def strTitle(func):
    def inner(): #to modify existing functions functionality
        str1 = func() #strname() is called and str="hi welcome "
        return str1.title()
    return inner

@strupper
def strname():
    return "hi welcome"

@strTitle
def myname():
    return "Madhav"

print(strupper(strname)())
print(strTitle(myname)())
print(strname())
print(myname())
```

```python
#adding decorators

def divdec(func): #second
    def inner(x,y): #third
        if y == 0:
            return "Dont give divisor as zero"
        else:
            return func(x,y)
    return inner #as it returns inner, inner() will be called (function passed as reference)

@divdec
def div(a,b): #1st step
    return a/b

print(div(60,0))
print(div(60,10))

```

## *Types of Methods*
- Instance method
- Class method
- Static method

```python
# instance method

class Subject:
    def __init__(self, Tamil, English, Maths):
        self.tamil=Tamil
        self.maths=Maths
        self.english=English

    def marks(self): #if we pass self as parameter then its called as instance method
        Total=self.tamil+self.english+self.maths
        return Total

roxxy = Subject(97,92,80)
roxxy.marks()
```

```python
class Demo():
    
    @classmethod
    def method(self,a,b):
        return "The Total is "+ str(a + b)

    @classmethod
    def method1(cls): #cls is a naming convention (not a rule ) #cls=class name here cls=demo
        return "Hi I\'m a class decorator"

obj = Demo()

print(obj.method(300,200))
print(obj.method1())

print('------------------')

print(Demo.method1())
print(Demo.method(40, 60))
```

```python
class Demo():
    var="Hi I\'m a class method"
    @classmethod
    def method(self,a,b):
        return "The Total is "+ str(a + b)

    @classmethod
    def method1(cls,x,y): #cls is a naming convention (not a rule ) #cls=class name here cls=demo
        print(x*y)
        return demo.var

obj = Demo()

print(obj.method(300,200))
print(obj.method1(100,200))

print('------------------')

print(Demo.method1(200,100))
print(Demo.method(40, 60))
```

```python
class demo :
    var="Hi I\'m a class method"

    def method(self,a,b):
        return "The total is "+str(a+b)

    def method1(cls,x,y): #cls is a naming convention (not a rule ) #cls=class name here cls=demo
        print(x*y)
        return demo.var
    @staticmethod
    def method2():
        return "Hi I\'am a static method"

obj = demo()

print(obj.method(3,5))
print(obj.method1(3,5))
print(obj.method2())
```
