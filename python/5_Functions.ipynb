## *Functions -(user defined)*
A function/method is a block of code which only runs when it is  called. 
You can pass data, known as parameter/arguments into a function. A  
function can return data as a result.

**Keyword**
- def
- return

**syntax**
- def \<function_name\>()

```python
#function Creation
def hello():
    print("Hello")
hello()#function calling
hello()
```

```python
def hello():
    print("Hello")
def Welcome():
    print("Welcome")
hello()
Welcome()
#function used for code reusability
```

## *function with parameters & arguments*

```python
def func(x,y):#x,y are parameters
    print("x =",x)
    print("y =",y)
    print("Total =",x+y)
func(10,20)#10,20 are arguments
func(100,100)
```

## *Defualt parameter*

```python
def func(x=100,y=100):#x,y are having default values - defualt parameters
    print("x =",x)
    print("y =",y)
    print("Total =",x+y)
print()
func()
func(10,20)#10,20 are arguments - default parameter are overwritten through default parameter
func(100,100)
func(200)
func(y=100)#argument to a specific parameter
```

## *Return*

```python
def func(x=100,y=100):
    return x+y
ans=func(100,50)#value can be stored in the variable using return
print("Ans =",ans)
print(func(100,200))#print doesn't store value as variable it just print and forgots
answer=func(func(100,100),func(200))
print("Answer =",answer)
```

```python
def add(x=0,y=0):
    return x+y
def sub(x=0,y=0):
    return x-y
x= add(100,100)
print("Add =",x)
y=sub(100,50)
print("Sub =",y)
ans = sub(add(100,100),sub(100,50))
print("Ans =",ans)
```

```python
#Task - Find Factorial
list1=[1,2,3,4,5]
list2=[1,2,3,4,5,6]
list3=[1,2,3,4,5,6,7,8]
def factorial(list):
    fact=1
    for num in list:
        fact=fact*num
    return fact
print(factorial(list1))
print(factorial(list2))
print(factorial([1,2,3,4,5,6,7]))
factorial(list3)
```

## *ARBITRARY ARGUMENTS*
If you do not know how many arguments that will be passed into  your 
function, add * before the parameter name in the function definition. This way 
the function will receive a tuple of arguments, and can access the items 
accordingly.

**Types of arguments**

- *args - argument stores in tuple
- **kwargs - keyworded argument stores in dictionary

```python
def func(*x):#*x - argument stores in tuple
    print(x,type(x))
func(10)
func("hi")
func(True)
func(50.55)
print("------------------")
func(10,"hi",True,50.50)
```

```python
def func(**x):#**x - keyworded argument stores in dictionary
    print(x,type(x))
func(a=10)
func(b="hi")
func(c=True)
func(d=50.55)
```

## *Function Overloading*
Function overloading or method overloading is the ability to create multiple functions of the same name with different implementations.
In Python you can define a method insuch a way that there are multiple ways to call it. Given a single method or function, we can specify the number  ofparameters yourself. Depending on the function definition, it can be called with zero, one, two or more parameters.

```python
def func(x):
    return x
def func(x,y):
    return x+y
def func(x,y,z):
    return x+y+z
print(func(20,40,60))
#here python overrides each function function overloading is not possible
```
