## *SELF PARAMETER*

The self parameter is a reference to the current instance of the  class, 
and is used to access variables that belongs to the class. It does  not have to 
be named self , you can call it whatever you like, but it has  to be the first 
parameter of any function in the class:

note : function inside class should have self as its first argument

```python
class Main:
    x=100
    y=200
    def func(self):
        print("function inside class")
        print(self)
obj = Main()
print(obj.x,obj.y)
obj.func()
print(obj)
```

```python
class Main:
    x=100
    y=200
    def func(self,a,b):
        print("function inside class")
        print(f"these are func variables : {a} , {b}")
obj = Main()
print(obj.x,obj.y)
obj.func(obj.x,obj.y)
```

## *CONSTRUCTOR - FUNCTION*
A constructor is a special kind of function/method that Python calls  
when it instantiates an object using the definitions found in your class.  Python 
relies on the constructor to perform tasks such as initializing  (assigning values 
to) any instance variables that the object will need  when it starts.

Syntax:

\__init\__\()

```python
class Main:
    def __init__(self,x,y):
        print("constructor calling")
        self.a =x
        self.b=y
obj = Main(300,3000)
print(obj.a,obj.b)
obj = Main(300,200)
print(obj.a,obj.b)
```

## *DESTRUCTORS - FUNCTION*
Destructors are called when an object gets destroyed. In Python, 
destructors are not needed as much needed in C++ because Python has a 
garbage collector that handles memory management automatically.

Syntax:

__del__()

```python
class Main:
    def __init__(self,x,y):
        print("constructor calling")
        self.a =x
        self.b=y
    def __del__():
        print("destructor calling")
obj = Main(300,3000)
print(obj.a,obj.b)
obj1 = Main(300,200)
print(obj1.a,obj1.b)
obj2 = Main(30,20)
print(obj2.a,obj2.b)
obj3 = Main(300000,200000)
print(obj3.a,obj3.b)
```

## *POLYMORPHISM*
The word polymorphism means having many forms. In simple  words, we 
can define polymorphism as the ability of a message to be  displayed in more 
than one form.
Real life example of polymorphism, a person at the same time can  
have different characteristic. Like a man at the same time is a father, a  
husband, an employee. So the same person posses different behavior  in 
different situations. This is called polymorphism

```python
class India:
    def capital(self):
        print("New delhi is the capital of India.")
    def language(self):
        print("Hindi The primary language of India.")
    def type(self):
        print("India is devalaping country")
class usa:
    def capital(self):
        print("Washington DC is the capital of USA.")
    def language(self):
        print("English The primary language of USA.")
    def type(self):
        print("USA is devalaped country")
obj_ind = India()
obj_usa = usa()
for country in [obj_ind,obj_usa]:
    country.capital()
    country.language()
    country.type()
    print()
```

## *ENCAPSULATION*
In normal terms Encapsulation is defined as wrapping up of data and  
information under a single unit. In Object Oriented Programming,  
Encapsulation is defined as binding together the data and the functions  that 
manipulates them.

```python
class xyz:
    x = 10 # pub Variable
    y = 20
obj = xyz()
print(obj.x,obj.y)
obj.x=1000
print(obj.x,obj.y)
```

```python
class xyz:
    __x = 10 # pvt variable
    __y = 20
    def getattr(self):
        print(f"This is my Class Variable are = {self.__x},{self.__y}")  
obj = xyz()
obj.getattr()
obj.x= 1000
obj.getattr()
```

## *DATA ABSTRACTION*
Data Abstraction - Abstraction means displaying only essential  
information and hiding the details. Data abstraction refers to providing  only 
essential information about the data to the outside world, hiding  the 
background details or implementation

```python
class Clothing(object):
    def __init__(self, type, color, size, price=None):
        self.type = type
        self.color = color
        self.size = size
        self.price = price
    def set_price(self, price):
        # Set the price of an item of clothing.
        self.price = price
        print(f"Setting the price of the {self.color} {self.type} to ${price}")
    def get_price(self):
        # Get the price of an item of clothing, if price is set.
        try:
            assert self.price != None
            print(f"The {self.color} {self.type} costs ${self.price}.")
        except:
            print(f"The price of the {self.color} {self.type} hasn't been set yet!")
    def promote(self, percentage):
        # Lower the price, if initial price is set.
        try:
            self.price = self.price * (1-percentage/100)
            print(f"The price of the {self.color} {self.type} has been reduced by {price}")
        except:
            print(f"Oops. Set an initial price first!")
bluejeans = Clothing("jeans", "blue", 12)
bluejeans.get_price()
bluejeans.set_price(100)
bluejeans.get_price()
bluejeans.promote(25)
bluejeans.get_price()
```

## *INHERITANCE*
Inheritance allows us to define a class that inherits all the methods  
and properties from another class. Parent class is the class being  inherited 
from, also called base class. Child class is the class that  inherits from another 
class, also called derived class.

## *Single inheritance:*
When a child class inherits from only one parent class, it is 
called as single inheritance

```python
class Parent:
    def P_ability(self):
        print("Can Sing")
class Child(Parent):
    def C_ability(self):
        print("can dance")
obj = Child()
obj.C_ability()
obj.P_ability()
```

## *Multiple inheritance:*
When a child class inherits from multiple parent classes,  it is 
called as multiple inheritanc

```python
class Add:
    def add(self,x,y):
        print("x + Y = ",x+y)
class Sub:
    def sub(self,x,y):
        print("x - Y = ",x-y) 
class Mul:
    def mul(self,x,y):
        print("x * Y = ",x*y)
        
class Div(Add,Sub,Mul):
    def div(self,x,y):
        print("x / Y = ",x/y)
calc = Div()
calc.add(5,5)
calc.sub(5,5)
calc.mul(5,5)
calc.div(5,5)
```

## *Multilevel inheritance:*
When we have child and grandchild relations

```python
class Animal:
    def speak(self):
        print("Animal Speaking")
class Dog(Animal):
    def bark(self):
        print("Dog Barking ")
class Lion(Dog):
    def roar(self):
        print("Lion Roaring")
obj = Lion()
obj.roar()
obj.bark()
obj.speak()
```

## *Hierarchical inheritance:*
More than one derived classes are created from a  
single base

```python
class Animal:
    def speak(self):
        print("Animal Speaking")
class Herbivorous(Animal):
    def bark(self):
        print("Dog Barking ")
class Omnivorous(Animal):
    def roar(self):
        print("Lion Roaring")
class Cat(Herbivorous):
    def meow(self):
        print("Cat meow ")
class Tiger(Omnivorous):
    def roaring(self):
        print("Tiger Roaring")
obj1=Cat()
obj1.bark()
obj1.meow()
print()
obj2=Tiger()
obj2.roar()
obj2.roaring()
```

## *Hybrid inheritance :*
Formed by combining the above types of inheritance

```python
class Animal:
    def speak(self):
        print("Animal Speaking")
class Herbivorous(Animal):
    def bark(self):
        print("Dog Barking ")
class Omnivorous(Animal):
    def roar(self):
        print("Lion Roaring")
class Cat(Herbivorous,Omnivorous):
    def meow(self):
        print("Cat meow ")
obj1=Cat()
obj1.bark()
obj1.meow()
obj1.roar()
obj1.speak()
        
```

## *MRO - METHOD RESOLUTION ORDER* 
Method resolution order defines the order in which the base classes are 
searched when executing a method. First, the method or attribute is searched 
within a class and then it follows the order we specified while inheriting. This 
order is also called Linearization of a class and set of rules are called 
MRO(Method Resolution Order)

```python
class Animal:
    def speak(self):
        print("Animal Speaking")
class Dog(Animal):
    def bark(self):
        print("Dog Barking ")
class Lion(Animal,Dog):# TypeError: Cannot create a consistent method resolution order (MRO) for bases Animal, Dog
    def roar(self):
        print("Lion Roaring")
obj = Lion()
obj.roar()
obj.bark()
obj.speak()
```

## *SUPER() - FUNCTION*
The super() builtin returns a proxy object (temporary object of the 
superclass) that allows us to access methods of the base class.

In Python, super() has two major use cases:

• Allows us to avoid using the base class name explicitly

• Working with Multiple Inheritance

```python
class Parent:
    def P_ability(self):
        print("Can Sing")
class Child(Parent):
    def P_ability(self):
        print("can dance")
obj = Child()
obj.P_ability()
```

```python
class Animal:
    def roar(self):
        print("Animal Speaking")
class Dog(Animal):
    def roar(self):
        print("Dog Barking ")
        super().roar()
class Lion(Dog):
    def roar(self):
        print("Lion Roaring")
        super().roar()
obj = Lion()
obj.roar()
```
