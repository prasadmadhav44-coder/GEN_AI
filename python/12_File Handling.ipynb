## *File Handling*
File handling is an important part of any web application. Python  has several functions for creating, reading, updating, and deleting files.

open() – function

• The key function for working with files in Python is the open()  function.

• The open() function takes two parameters; filename, and mode.

• There are four different methods (modes) for opening a file:

TYPE OF OPENING A FILE

- "r" - Read - a file for reading, error if the file does not exist
- "a" - Append - Opens a file for appending, creates the file if it does not  exist
- "w" - Write - Opens a file for overwriting, creates the file if it does not  exist
- "x" – Create- Creates the Default value. Opens e specified file, returns an error 
if the file exists

## *read - 'r'*

```python
myfile=open(r"C:\GEN_AI\Dummy\dummy.txt","r")

print(myfile.tell())

content=myfile.read(100)
print(content)
print(myfile.tell())

content=myfile.read(200)
print(content)
print(myfile.tell())

myfile.seek(0)
content=myfile.read(200)
print(content)
print(myfile.tell())

myfile.close()
```

### *with keyword*

```python
with open(r"C:\GEN_AI\Dummy\dummy.txt","r") as myfile:
    content=myfile.readline()
    print(content)
    content=myfile.readlines()
    print(content)
    print("------------------------")
    print(content[0])
    print(content[1])
    for line in content:
        print(line)
```

## *'a' - append*

```python
with open(r"C:\GEN_AI\Dummy\dummy.txt","a") as myfile:
    myfile.write("\nHello, World!")
```

## *'w' - write*

```python
with open(r"C:\GEN_AI\Dummy\dummy1.txt","w") as myfile:
    write_content="This is a new content of the file."
    myfile.write(write_content)
```

```python
with open(r"C:\GEN_AI\Dummy\dummy1.txt","a") as myfile:
    myfile.write("\nNew line added to the file.")
```

## *'x'-create*

```python
with open(r"C:\GEN_AI\Dummy\dummy1.txt","x") as myfile:
    write_content="This is a new content of the file."
    myfile.write(write_content)
```

```python
with open(r"C:\GEN_AI\Dummy\dummy2.txt","a") as myfile:
    write_content="This is a new content of the file."
    myfile.write(write_content)
```

```python
with open(r"C:\GEN_AI\Dummy\dummy.txt", "r") as myfile:
    for i in range(14):
        print(myfile.read(50))
```

```python
with open(r"C:\GEN_AI\Dummy\dummy.txt", "r") as myfile:
    while True:
        content =  myfile.read(50)
        if content == "":
            break
        print(content)
```
