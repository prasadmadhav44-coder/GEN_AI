```python
# Interview   
# check palindrome
# 1 Find weather the String is palindrome
# 1.  Input: racecar
#     Output: True
# 2.  Input: MalaYa lAm
#     Output: True
# 3   Input: Sir, I demand, I am a maid named Iris
#     Output: True
# 4   Input : Red roses run no risk. Sir, on Nurse's Order.
```

```python
str1="racecar"
rev=str1[::-1]
if str1==rev:
    print(True)
else:
    print(False)
```

```python
str1="MalaYa lAm"
str2=str1.lower().replace(' ','')
rev=str2[::-1]
if str2==rev:
    print(True)
else:
    print(False)
```

```python
str1="Sir, I demand, I am a maid named Iris"
str2=str1.lower().replace(' ','').replace(',','')
rev=str2[::-1]
if str2==rev:
    print(True)
else:
    print(False)
```

```python
str1="Red roses run no risk. Sir, on Nurse's Order."
str2=""
for ch in str1:
    if ch.isalpha():
        str2=str2 + ch.lower()
if str2==str2[::-1]:
    print(True)
else:
    print(False)
```

```python
def check_palindrome(str1):
    str2=""
    for ch in str1:
        if ch.isalpha():
            str2=str2 + ch.lower()
    if str2==str2[::-1]:
        return True
    else:
        return False
print(check_palindrome("racecar"))
print(check_palindrome("Sir, I demand, I am a maid named Iris"))
print(check_palindrome("Red roses run no risk. Sir, on Nurse's Order."))
```

```python
def Cart_free(mystr):
    newstr =''
    mydict = {}
    for each in mystr:
        if each.isalnum():
            newstr = newstr + each.lower()
    for each_1 in newstr:
        if each_1 not in mydict.keys():
            mydict[each_1]=1
        else:
            mydict[each_1]+= 1
    return mydict
ans = Cart_free("Hello WorLD")
print(ans)
print(Cart_free("MANJUNATHAN"))
print(Cart_free("9840243168"))
print(Cart_free("MalaYa lAm"))
```
