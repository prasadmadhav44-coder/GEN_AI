## *List Functions*

```python
num = [1,2,3,4,5,6,7,7,7]
list.append(7)# Adds an element at the end of the list
print(list)
num.pop(2)#Removes the element at the end of the list
print(num)
num.insert(2,3)# Adds an element at the specified position
print(num)
num.remove(3)# Removes the item with the specified value
print(num)
num.clear()#Removes all the elements from the list
print(num)
```

```python
num = [1,2,9,4,5,6,7,6,7]
print(num.index(4))# Returns the index of the first element with the specified value
print(num.count(7))# Returns the number of elements with the specified value
num.sort()
print(num)#Sorts the list
num.reverse()
print(num)#Reverses the order of the list
num1=[2,3,4,5]
num.extend(num1)
print(num)#Add the elements of a list (or any iterable), to the end of the current list
print(num1.copy())
```

## *Tuple Functions*

```python
tup = (1,2,3,4,4,4,5)
tup.count(4)
print(tup)
tup.index(4)
```

## *Set Functions*

```python
set1={1,2,3,4,5,6}
set2={5,4,3,2,1}
print(set1.union(set2))
print(set1.intersection(set2))
print(set1.difference(set2))
```

## *List Comprehension*

```python
names = ["alice", "bob", "charlie"]
uppercase_names = [name.upper() for name in names]
print(uppercase_names)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = [n for n in numbers if n % 2 == 0]
print(evens)
```

```python
numbers = [1, 2, 3, 4]
labels = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print(labels)
```

```python
matrix = [[1, 2], [3, 4]]
flat = [num for row in matrix for num in row]
print(flat)
```
