## *Lambda Function*
- A lambda function is a small anonymous function.
- A lambda function can take any number of arguments, but can only  have 
one expression.
- called as nameless function
Syntax:

funcname = lambda arguments : expression

```python
myfunc=lambda x,y: x*y
myfunc(5,6)
```

## *map()*

Syntax:
map(funcname,iterable)


```python
#i/p = [1,2,3,4,5]
#o/p = [1,4,9,16,25]
print("Squared list:", list(map(lambda x: x*x, [1,2,3,4,5])))
print("Cubed list:", list(map(lambda x: x*x*x, [1,2,3,4,5])))
print("Squared tuple:", tuple(map(lambda x: x*x, (1,2,3,4,5))))
print("Cubed tuple:", tuple(map(lambda x: x*x*x, (1,2,3,4,5))))
print("Squared set:", set(map(lambda x: x*x, {1,2,3,4,5})))
print("Cubed set:", set(map(lambda x: x*x*x, {1,2,3,4,5})))
print("Squared dict:", dict(map(lambda x: (x,x*x), {1,2,3,4,5})))
```

## *Filter()*
Syntax:
filter(funcname,iterable)

```python
print("Filtered even list:", list(filter(lambda x: x%2==0, [1,2,3,4,5])))
print("Filtered odd list:", list(filter(lambda x: x%2!=0, [1,2,3,4,5])))
print("Filtered even tuple:", tuple(filter(lambda x: x%2==0, (1,2,3,4,5))))
print("Filtered odd tuple:", tuple(filter(lambda x: x%2!=0, (1,2,3,4,5))))
print("Filtered even set:", set(filter(lambda x: x%2==0, {1,2,3,4,5})))
print("Filtered odd set:", set(filter(lambda x: x%2!=0, {1,2,3,4,5})))
print("Filtered even dict:", dict(filter(lambda x: x[0]%2==0, {1:1,2:4,3:9,4:16,5:25}.items())))
print("Filtered odd dict:", dict(filter(lambda x: x[0]%2!=0, {1:1,2:4,3:9,4:16,5:25}.items())))
```

## *GENERATOR FUNCTION*

A generator-function is defined like a normal function, but  whenever it needs 
to generate a value, it does so with the yield  keyword rather than return. If the 
body of a def contains yield, the  function automatically becomes a 
generator function.

```python
def myfunc(x,y):
    yield x*y
    yield x+y
    yield x-y
print(list(myfunc(5,6)))
for i in myfunc(5,6):
    print(i)
```

## *Scope of the variable*

Scope refers to the coding region from which particular Python object  is accessible. Hence one cannot access any particular object from  anywhere from the code, the accessing has to be allowed by the scope of the object.

Types of scope

- Local scope
- Global scope

```python
#local variable
def myfunc():
    x = 10
    y = 10
    print("local",x)
    print("local",y)
myfunc()
print("global",x)
print("global",y)
```

```python
#global variable
x = 10
def myfunc():
    global y
    y = 20
    print("local",y)
    print("local",x)
print("global",x)
# print("global",y)
myfunc()
print("global",x)
print("global",y)
```

## *RECURSION*
A function is called within the same function is called Recursion. 
Recursion is  a common mathematical and programming concept. It means 
that a  function calls itself. This has the benefit of meaning that you can loop  
through data to reach a result.
The developer should be very careful with recursion as it can be  quite 
easy to slip into writing a function which never terminates, or one  that uses 
excess amounts of memory or processor power. However,  when written 
correctly recursion can be a very efficient and  mathematically-elegant 
approach to programming.


```python
def factorial(n):
    return 1 if n == 0 else n * factorial(n - 1)
factor = factorial(5)
print(factor)
```

```python
def fibonacci(n):
    return n if n <= 1 else fibonacci(n - 1) + fibonacci(n - 2)
fib = fibonacci(10)
print(fib)
```
