## *Regex*

```python
import re
with open(r"C:\GEN_AI\Dummy\dummy.txt","r") as myfile:
    pattern=r"Hello"
    content=myfile.read()
    matches=re.findall(pattern,content)#findall() returns a list of all occurrences of the pattern in the string.
    print(matches)
    print(len(matches))
    split_content=re.split(" ",content)#split() returns a list of strings that are split by the pattern.
    for i in split_content:
        print(i)
    replace_content=re.sub(pattern,"Hi",content)#sub() returns a string with all occurrences of the pattern replaced by the replacement string.
    print(replace_content)
    search_result=re.search(pattern,content)   #search() returns a match object if the pattern is found in the string, otherwise it returns None.
    print(search_result)
```

## *Import - OS*
The os module allows us to access functionality of the underlying operating 
system. So we can perform tasks such as: navigate the file system, obtain file 
information, rename files, search directory trees, fetch environment variables, 
and many other operations.

```python
import os
print (os.getcwd())#getcwd() returns the current working directory.

x=open(r"C:\GEN_AI\Dummy\dummy.txt","r")#open() opens a file and returns a file object.
print(x.read())#read() reads the entire file and returns it as a string.
x.close()#close() closes the file.
```

```python
import os
print("Current working directory:", os.getcwd())
os.chdir(r"C:\GEN_AI\Dummy")#chdir() changes the current working directory to the specified path.
print("Working directory changed to:", os.getcwd())
print("List of files and directories in the current working directory:")#listdir() returns a list of all files and directories in the specified path.
mylist=os.listdir()
for i in mylist:
    if i.endswith(".txt"):#endswith() returns True if the string ends with the specified suffix, otherwise it returns False.
        print(i)
```
