2. Write a Python program to find the factorial of a given non-negative integer.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

factorial(5)
```

3. Write a Python program to reverse a given integer without using string conversion or slicing.

```python
n = 1234
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print( rev)
```

4. Write a program to compare two numbers using relational operators.

5. Write a Python program to check whether a given number is divisible by both 4 and 5.

```python
def divByfourandfive(num):
    if num % 4 == 0 and num % 5 == 0:
        return True
    else:
        return False
    
divByfourandfive(20)
```
