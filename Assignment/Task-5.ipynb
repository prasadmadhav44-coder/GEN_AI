## *Pattern*

```python
#For Loop
row=int(input("Enter a number between (0-9) :"))
for i in range(row):
    for j in range(1,i+2):
        print(i+1,end='')
    print()
```

```python
#While Loop
row = int(input("Enter no. of rows: "))
a = 1
while a <= row:
    b = 1
    while b <= a:
        print(a, end='')
        b += 1
    print()
    a += 1
```
