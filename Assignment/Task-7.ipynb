## *Python Assessment Questions Part -03*

### 1.Write a Python program to find the length of a list using len().

```python
list = [1,2,3,4,5,6]
print(len(list))
```

### 2.Write a Python program to find the maximum and minimum value in a list using max() and min().


```python
list=[1, 2, 3, 4, 5, 6]
print("The maximum value in the list is:", max(list))
print("The minimum value in the list is:", min(list))
```

### 3.Use sum() to calculate the total of all numbers in a list.

```python
list=[1, 2, 3, 4, 5, 6]
print("The sum of the values in the list is:", sum(list))
```

### 4.Create a list of numbers and sort them in ascending and descending order using sort()

```python
list=[33, 12, 5, 8, 19]
print("The list in ascending order is:", sorted(list))
print("The list in descending order is:", sorted(list, reverse=True))
```

### 5.Write a Python program to create a tuple and print it.

```python
tuple = (1, 2, 3, 4, 5)
print(tuple)
```

### 6.Write a program to access the first, middle, and last elements of a tuple using indexing

```python
tuple = (1, 2, 3, 4, 5)
print(tuple[0])
print(tuple[len(tuple)//2])
print(tuple[-1])
```

### 7.Write a program to slice a tuple (first 3 elements, last 3 elements).


```python
tuple = (1, 2, 3, 4, 5, 6)
print("first half of the tuple:", tuple[:len(tuple)//2])
print("second half of the tuple:", tuple[len(tuple)//2:])
```

### 8.Write a program to remove an element from a set using discard()

```python
set = {1, 2, 3, 4, 5}
set.discard(3)
print(set)
```

### 10.Write a program to update the value of an existing key in a dictionary.

```python
dictionary = {'a': 1, 'b': 2, 'c': 3}
dictionary['c'] = 4
print(dictionary)
```

### 11.Write a program to remove the last inserted key-value pair using popitem().

```python
dictionary = {'a': 1, 'b': 2, 'c': 3}
dictionary.popitem()
print(dictionary)
```

### 12.Write a program to remove all key-value pairs using clear().

```python
dictionary = {'a': 1, 'b': 2, 'c': 3}
dictionary.clear()
print(dictionary)
```

### 13.Write a program to get all keys using keys().

```python
dictionary = {'a': 1, 'b': 2, 'c': 3}
for key in dictionary.keys():
    print(key)
    
```

### 14.Write a program to get all key-value pairs using items().

```python
dictionary = {'a': 1, 'b': 2, 'c': 3}
for item in dictionary.items():
    print(item)
    
```

### 15.Write a program to check if a key exists in a dictionary using in.


```python
dictionary = {'a': 1, 'b': 2, 'c': 3}
key=input("Enter a key to check if it exists in the dictionary: ")
if key in dictionary:
    print(f"Key '{key}' exists in the dictionary.")
else:
    print(f"Key '{key}' does not exist in the dictionary.")
```
