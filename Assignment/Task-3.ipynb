## *Loop And Functions*

### 1. Print numbers from 1 to 10 using a for loop

```python
for i in range(1,11):
    print(i,end=' ')
```

### 2. Print numbers from 10 to 1 using a while loop. 

```python
n=1
while n<=10:
    print(n,end=' ')
    n+=1
```

### 3. Write a program to print the multiplication table of a given number. 

```python
num=int(input("Give a Number :"))
for i in range(1,11):
    print(num," * ",i," = ",num * i)
    
```

### 4. Write a program to find the sum of first N natural numbers using a loop. 

```python
N=int(input("Enter a Number :"))
total=0
for i in range(1,N+1):
    total+=i
print(total)
```

### 5. Print the following pattern using a loop: blow pattern

```python
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
row=int(input("Enter no.of.rows :"))
for i in range(row):
    for j in range(0,i+1):
        print(Symbol,end='')
    print()
```

### 6. Write a function greet(name) that prints "Hello, \<name\>"

```python
def greet(name):
    print("Hello, ",name)
greet("Madhav")
```

### 7. Write a function add_numbers(a, b) that returns the sum of two numbers. 

```python
def add_numbers(a,b):
    return a+b
add_numbers(10,10)
```

### 8. Write a function is_even(n) that returns True if the number is even, otherwise False. 

```python
def is_even(n):
    if n%2==0:
        return True
    else:
        return False
num =  int(input("Enter a Number:"))
is_even(num)
```

### 9. Write a function square(n) that returns the square of a given number. 

```python
def square(n):
    return n ** 2
num =  int(input("Enter a Number:"))
square(num)
```

### 10. Write a function max_of_two(a, b) that returns the larger of two numbers 

```python
def max_of_two(a,b):
    if a > b:
        return a
    else:
        return b
max_of_two(8,6)
```
