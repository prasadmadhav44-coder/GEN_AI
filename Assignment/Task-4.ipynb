## *Pattern*

```python
#For Loop
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
row=int(input("Enter no.of.rows :"))
for i in range(row,0,-1):
    for _ in range(i):
        print(Symbol,end='')
    print()
```

```python
#While loop
Symbol = input("Give any Symbol: (!/@/#/$/%/^/&): ")
row = int(input("Enter no. of rows: "))
a = 1
while a <= row:
    b = 1
    while b <= row+1-a:
        print(Symbol, end='')
        b += 1
    print()
    a += 1
```
