### 5.Write a python program to check if a number is even or odd using the modulus operator.


```python
number=int(input("Enter a number: "))
if number%2==0:
    print(number, "is even")
else:
    print(number, "is odd")
```

### 6. if a=15 and b=4, find the sum, difference, product, division, and modulus.

```python
a=int(input("Enter a: "))
b=int(input("Enter b: "))
if a==15 and b==4:
    print("Sum:", a+b)
    print("Difference:", a-b)
    print("Product:", a*b)
    print("Division:", a/b)
    print("Modulus:", a%b)
```

### 7.write a python program that checks if a number is positive, negative, or zero. 

```python
number=int(input("Enter a number: "))
if number>0:
    print(number, "is positive")
elif number<0:
    print(number, "is negative")
else:
    print("The number is zero")
```
